```python
from pathlib import Path
import pandas as pd
  
DATASETS_DIR = Path(r"C:\\Users\\goku\\Documents\\NFL_ML_Predictions\\backend\\data\\datasets")

def load_latest_by_filename_date(
    datasets_dir: Path = DATASETS_DIR,
    prefix: str = "game_features_",
) -> tuple[pd.DataFrame, Path]:
    """
    Load the CSV with the largest YYYYMMDD date found in its filename.

    Expected filename format:
      game_features_YYYYMMDD.csv

    Why this works well:
    - Ignores accidental re-saves / modified times
    - Picks the newest dataset by its intended date stamp
    """
    # Example match: game_features_20260103.csv
    files = list(datasets_dir.glob(f"{prefix}*.csv"))

    if not files:
        raise FileNotFoundError(f"No files matched '{prefix}*.csv' in: {datasets_dir}")

    def extract_yyyymmdd(path: Path) -> int:
        """
        Pull YYYYMMDD from 'game_features_YYYYMMDD.csv' -> int(YYYYMMDD).
        If the filename doesn't match, return -1 so it won't win.
        """
        # path.stem is filename without extension: "game_features_20260103"
        suffix = path.stem.replace(prefix, "")

        # Only accept exactly 8 digits for YYYYMMDD
        if len(suffix) == 8 and suffix.isdigit():
            return int(suffix)

        return -1

    # Find the file with the largest date stamp
    latest_file = max(files, key=extract_yyyymmdd)

    # Guard: if all files were invalid format, max() would return one with -1
    if extract_yyyymmdd(latest_file) == -1:
        raise ValueError(
            f"Found CSVs, but none matched '{prefix}YYYYMMDD.csv' in: {datasets_dir}"
        )

    df = pd.read_csv(latest_file)
    df = df.sort_values(["season", "week"], ascending=[False, False])
    return df, latest_file


df, path = load_latest_by_filename_date()
print("Loaded:", path)
print(df.head())
```
