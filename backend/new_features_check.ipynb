```python
import pandas as pd
from typing import List

df = pd.read_csv('C:/Users/iProg/OneDrive/Documents/Football_predict/nfl_prediction_system/NFL_ML_Predictions/backend/data/Nfl_data_sorted.csv')
```

```python
df.infer_objects()
df.info()
print(df.describe(include='all'))
print(df.columns)
```

```python
for home, away in zip(df['home_team'].unique(), df['away_team'].unique()):
  # check how many times the teams have played each other and who won more
  matchup = df[(df['home_team'] == home) & (df['away_team'] == away)]
  print("Matchup between", home, "and", away, ":", len(matchup), matchup)
  if not matchup.empty:
    home_counts = matchup[['home_team','winner']].value_counts()
    away_counts = matchup[['away_team','winner']].value_counts()
    print("Home counts:", home_counts)
    print("Away counts:", away_counts)
    df['rivalry'] = home_counts.get((home, home), 0) - away_counts.get((away, away), 0)
    print("Rivalry column added for", home, "vs", away, "with values:", df['rivalry'].values)
  else:
    df['rivalry'] = 0
```
